## Atividade 1
Escreva o método das diferenças finitas dentro do arquivo $utils/algoritmos.py$.

````python
def dp(f, x, h):
    return (f(x + h) - f(x)) / h


def dr(f, x, h):
    return (f(x) - f(x - h)) / h


def dc(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
````

Use os esquemas numéricos de **diferença finita regressiva de ordem 1**, **diferença finita progressiva de ordem 1** e **diferença finita central de ordem 2** para aproximar as seguintes derivadas:

- $f'(x)$ onde $f(x) = \sin(x)$ e $x = 2$  
- $f'(x)$ onde $f(x) = e^{-x}$ e $x = 1$

Use $h = 10^{-2}$ e $h = 10^{-3}$ e compare com os valores obtidos pela derivada exata.

**Respostas esperadas:**

- Para $f'(x)$ onde $f(x) = \sin(x)$ e $x = 2$:

  - Progressiva (ordem 1): $-0.42069$ e $-0.41660$
  - Regressiva (ordem 1): $-0.41159$ e $-0.41569$
  - Central (ordem 2): $-0.41614$ e $-0.41615$
  - Exata: $\cos(2) = -0.41615$

- Para $f'(x)$ onde $f(x) = e^{-x}$ e $x = 1$:

  - Progressiva (ordem 1): $-0.36605$ e $-0.36788$
  - Regressiva (ordem 1): $-0.36972$ e $-0.36806$
  - Central (ordem 2): $-0.36789$ e $-0.36788$
  - Exata: $-e^{-1} = -0.36788$

````python
import sys
import os
import numpy as np
from scipy.differentiate import derivative

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    dp,
    dr,
    dc,
)

def main():
    h1 = 1e-2
    h2 = 1e-3
    eps = np.finfo(float).eps

    print("f(x) = sin(x)")

    f = lambda x: np.sin(x)

    print("Diferença progressiva:")
    print("h = 0.01 | f'(x)=", dp(f, 2, h1))
    print("h = 0.001 | f'(x)=", dp(f, 2, h2))
    print("h = eps | f'(x)=", dp(f, 2, np.sqrt(eps)))

    print("Diferença regressiva:")
    print("h = 0.01 | f'(x)=", dr(f, 2, h1))
    print("h = 0.001 | f'(x)=", dr(f, 2, h2))
    print("h = eps | f'(x)=", dr(f, 2, np.sqrt(eps)))

    print("Diferença central:")
    print("h = 0.01 | f'(x)=", dc(f, 2, h1))
    print("h = 0.001 | f'(x)=", dc(f, 2, h2))
    print("h = eps | f'(x)=", dc(f, 2, eps ** (1 / 3)))

    print("Derivada com SciPy:")
    r = derivative(f, 2)
    print("f'(x)=", r.df)

    fd = lambda x: np.cos(x)
    print("Derivada analítica:")
    print("f'(x) = cos(x)")
    print("f'(x)=", fd(2))

    print("f(x) = e^(-x)")

    f = lambda x: np.exp(-x)

    print("Diferença progressiva:")
    print("h = 0.01 | f'(x)=", dp(f, 1, h1))
    print("h = 0.001 | f'(x)=", dp(f, 1, h2))
    print("h = eps | f'(x)=", dp(f, 1, np.sqrt(eps)))

    print("Diferença regressiva:")
    print("h = 0.01 | f'(x)=", dr(f, 1, h1))
    print("h = 0.001 | f'(x)=", dr(f, 1, h2))
    print("h = eps | f'(x)=", dr(f, 1, np.sqrt(eps)))

    print("Diferença central:")
    print("h = 0.01 | f'(x)=", dc(f, 1, h1))
    print("h = 0.001 | f'(x)=", dc(f, 1, h2))
    print("h = eps | f'(x)=", dc(f, 1, eps ** (1 / 3)))

    print("Derivada com SciPy:")
    r = derivative(f, 1)
    print("f'(x)=", r.df)

    fd = lambda x: -np.exp(-x)
    print("Derivada analítica:")
    print("f'(x) = e^(-x)")
    print("f'(x)=", fd(1))


if __name__ == "__main__":
    main()
````

## Atividade 2
As tensões na entrada $v_i$ e saída $v_o$ de um amplificador foram medidas em regime estacionário conforme a tabela abaixo:

| $v_i$ (V) | 0.0 | 0.5 | 1.0 | 1.5 | 2.0 | 2.5 | 3.0 | 3.5 | 4.0 | 4.5 | 5.0 |
|:----------:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| $v_o$ (V) | 0.0 | 1.05 | 1.83 | 2.69 | 3.83 | 4.56 | 5.49 | 6.56 | 6.11 | 7.06 | 8.29 |

O ganho é definido por:

$$
G = \frac{\partial v_o}{\partial v_i}
$$

Calcule o ganho quando $v_i = 1$ e $v_i = 4.5$ usando as seguintes técnicas:

a. Derivada primeira numérica de **primeira ordem** usando o próprio ponto e o **próximo**.  
b. Derivada primeira numérica de **primeira ordem** usando o próprio ponto e o **anterior**.  
c. Derivada primeira numérica de **segunda ordem** usando o ponto **anterior e o próximo**.  
d. Derivada primeira analítica da função do tipo  
   $v_o = a_1 v_i + a_3 v_i^3$  
   ajustada aos pontos pelo critério dos **mínimos quadrados**.

**Respostas esperadas:**

| Caso | a | b | c | d |
|:-----|:--:|:--:|:--:|:--:|
| $v_i = 1$ | 1.72 | 1.56 | 1.64 | 1.86 |
| $v_i = 4.5$ | 2.46 | 1.90 | 2.18 | 1.14 |

````python
import sys
import os
import numpy as np

sys.path.append(os.path.abspath(os.path.join('..')))

try:
    from utils.algoritmos import regressao
except ImportError:
    def regressao(x, y, v):
        V = v(x)
        Vt = V.T
        return np.linalg.inv(Vt @ V) @ (Vt @ y)

def main():
    vi = np.array([0.0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0], dtype=float)
    vo = np.array([0.0, 1.05, 1.83, 2.69, 3.83, 4.56, 5.49, 6.56, 6.11, 7.06, 8.29], dtype=float)
    h = 0.5

    # Índices para vi = 1.0 e vi = 4.5
    idx_1 = np.where(np.isclose(vi, 1.0))[0][0]
    idx_45 = np.where(np.isclose(vi, 4.5))[0][0]

    # a. Diferença progressiva (ordem 1): (vo[i+1] - vo[i]) / h
    ga_1 = (vo[idx_1 + 1] - vo[idx_1]) / h
    ga_45 = (vo[idx_45 + 1] - vo[idx_45]) / h

    # b. Diferença regressiva (ordem 1): (vo[i] - vo[i-1]) / h
    gb_1 = (vo[idx_1] - vo[idx_1 - 1]) / h
    gb_45 = (vo[idx_45] - vo[idx_45 - 1]) / h

    # c. Diferença central (ordem 2): (vo[i+1] - vo[i-1]) / (2 * h)
    gc_1 = (vo[idx_1 + 1] - vo[idx_1 - 1]) / (2 * h)
    gc_45 = (vo[idx_45 + 1] - vo[idx_45 - 1]) / (2 * h)

    # d. Derivada analítica de vo = a1 * vi + a3 * vi^3 ajustada por mínimos quadrados
    v_func = lambda x: np.column_stack((x, x**3))
    A = regressao(vi, vo, v_func)
    a1, a3 = A[0], A[1]
    # Derivada analítica: dvo/dvi = a1 + 3 * a3 * vi^2
    gd_1 = a1 + 3 * a3 * (1.0**2)
    gd_45 = a1 + 3 * a3 * (4.5**2)

    print(f"Ajuste vo = a1*vi + a3*vi^3: a1 = {a1:.6f}, a3 = {a3:.6f}")
    print("\nTabela de Resultados do Ganho G = dvo/dvi:")
    print("-" * 55)
    print(f"{'Caso':<15} | {'a':>8} | {'b':>8} | {'c':>8} | {'d':>8} |")
    print("-" * 55)
    print(f"{'vi = 1.0':<15} | {ga_1:>8.2f} | {gb_1:>8.2f} | {gc_1:>8.2f} | {gd_1:>8.2f} |")
    print(f"{'vi = 4.5':<15} | {ga_45:>8.2f} | {gb_45:>8.2f} | {gc_45:>8.2f} | {gd_45:>8.2f} |")
    print("-" * 55)

if __name__ == "__main__":
    main()
````
