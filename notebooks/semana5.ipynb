## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import math
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import pontofixo, newton_raphson, secante

f1 = lambda x: math.exp(x) - x - 2
g1 = lambda x: math.exp(x) - 2

raiz_ponto_fixo, iteracoes = pontofixo(-1.8, g1)
print(f"ponto fixo = {raiz_ponto_fixo:.10f}, iterações={iteracoes}")

raiz_newton, iteracoes = newton_raphson(
    -1.8, f1, df=lambda x: math.exp(x) - 1
)
print(f"Newton-Raphson = {raiz_newton:.10f}, iterações={iteracoes}")

raiz_newton_aprox, iteracoes = newton_raphson(-1.8, f1)
print(f"Newton-Raphson (derivada aproximada) = {raiz_newton_aprox:.10f}, iterações={iteracoes}")

raiz_secante, iteracoes = secante(-1.8, -1.7, f1)
print(f"secante = {raiz_secante:.10f}, iterações={iteracoes}")
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import math


def f2(x):
    return math.cos(x) - x**2


def g2(x):
    # Forma iterativa fornecida no enunciado.
    return x + f2(x) / (math.sin(x) + 2 * x)


raiz, iteracoes = pontofixo(1.0, g2, TOL=5e-6)
print(f"raiz = {raiz:.8f}, f(raiz)={f2(raiz):.3e}, iterações={iteracoes}")
print(f"quinto dígito significativo: {raiz:.5f}")
assert round(raiz, 5) == 0.82413
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import math


def f3(x):
    return math.exp(-x**2) - 2 * x


def g3(x):
    return math.exp(-x**2) / 2


raiz_ponto_fixo, iteracoes = pontofixo(0.4, g3, TOL=1e-10)
raiz_newton, iteracoes_newton = newton_raphson(0.4, f3, TOL=1e-10)
raiz_secante, iteracoes_secante = secante(0.3, 0.5, f3, TOL=1e-10)

print(f"ponto fixo = {raiz_ponto_fixo:.10f}, iterações={iteracoes}")
print(f"Newton-Raphson = {raiz_newton:.10f}, iterações={iteracoes_newton}")
print(f"secante = {raiz_secante:.10f}, iterações={iteracoes_secante}")
print(f"resíduo = {f3(raiz_ponto_fixo):.3e}")
assert abs(raiz_ponto_fixo - 0.4193648) < 1e-6
````

## Atividade 4
Resolva os três últimos exercícios da semana anterior usando os métodos do ponto fixo.

````python
import math


# Exercício 4 da semana 4: circuito não linear com diodo.
thermal_voltage = 1.380649e-23 * 300.0 / 1.602176634e-19
reverse_current = 1e-12


def diode_equation(vd, voltage, resistance):
    try:
        diode_current = reverse_current * math.expm1(vd / thermal_voltage)
    except OverflowError:
        diode_current = math.inf
    return resistance * diode_current + vd - voltage


def diode_derivative(vd, resistance):
    try:
        exponential = math.exp(vd / thermal_voltage)
    except OverflowError:
        exponential = math.inf
    return resistance * reverse_current * exponential / thermal_voltage + 1


def solve_diode(voltage, resistance):
    initial = voltage if voltage <= 0.7 else 0.7
    newton = newton_raphson(
        initial,
        lambda vd: diode_equation(vd, voltage, resistance),
        TOL=1e-12,
        df=lambda vd: diode_derivative(vd, resistance),
        iter=100,
    )
    secant_result = secante(
        initial * 0.9,
        initial * 1.1 if initial else 0.1,
        lambda vd: diode_equation(vd, voltage, resistance),
        TOL=1e-12,
        iter=100,
    )
    return newton, secant_result


casos = [(30.0, 1e3), (3.0, 1e3), (3.0, 10e3),
         (0.3, 1e3), (-0.3, 1e3), (-30.0, 1e3), (-30.0, 10e3)]
for voltage, resistance in casos:
    raiz_newton, raiz_secante = solve_diode(voltage, resistance)
    print(
        f"diodo V={voltage:g} V, R={resistance / 1e3:g} kOhm -> "
        f"Newton={raiz_newton[0]:.3g} V, secante={raiz_secante[0]:.3g} V"
    )


# Exercício 5 da semana 4: comprimento da catenária.
distance = 500.0
maximum_sag = 50.0


def catenary_equation(cable_length):
    argument = distance / (2 * cable_length)
    return cable_length * (math.cosh(argument) - 1) - maximum_sag


def catenary_derivative(cable_length):
    argument = distance / (2 * cable_length)
    return math.cosh(argument) - 1 - argument * math.sinh(argument)


catenary_newton = newton_raphson(
    600.0, catenary_equation, TOL=1e-10, df=catenary_derivative
)
catenary_secant = secante(500.0, 800.0, catenary_equation, TOL=1e-10)
print(
    f"catenária -> Newton={catenary_newton[0]:.4f} m, "
    f"secante={catenary_secant[0]:.4f} m"
)
assert abs(catenary_newton[0] - 633.1621) < 1e-3


# Exercício 6 da semana 4: retificador de meia onda.
frequency = 1e3
inductance = 100e-3
resistance = 1e3
tan_phi = 2 * math.pi * frequency * inductance / resistance
phi = math.atan(tan_phi)


def current_equation(beta):
    return math.sin(beta - phi) + math.sin(phi) * math.exp(-beta / tan_phi)


def current_derivative(beta):
    return math.cos(beta - phi) - math.sin(phi) * math.exp(-beta / tan_phi) / tan_phi


beta_newton = newton_raphson(
    3.8, current_equation, TOL=1e-10, df=current_derivative
)
beta_secant = secante(3.6, 4.0, current_equation, TOL=1e-10)
print(
    f"retificador -> Newton={math.degrees(beta_newton[0]):.4f} graus, "
    f"secante={math.degrees(beta_secant[0]):.4f} graus"
)
assert abs(current_equation(beta_newton[0])) < 1e-9
assert abs(math.degrees(beta_newton[0]) - 212.2284) < 1e-2
````
