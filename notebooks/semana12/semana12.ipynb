## Atividade 1
Escreva os métodos regra de quadratura pelo ponto médio, Trapézio e Simpson dentro do arquivo $utils/algoritmos.py$.

````python
def medio(f, a, b):
    h = b - a
    return h * f((a + b) / 2)


def trapezio(f, a, b):
    h = b - a
    return h * (0.5 * f(a) + 0.5 * f(b))


def simpson(f, a, b):
    h = (b - a) / 2
    return h * ((1 / 3) * f(a) + (4 / 3) * f((a + b) / 2) + (1 / 3) * f(b))


def integral(metodo, f, a, b, n=1e-3):
    s = 0.0
    c = a
    d = a + n
    while d <= b:
        s += metodo(f, c, d)
        c = d
        d += n
    return s
````

## Atividade 2
Calcule numericamente o valor de  

$\displaystyle \int_{2}^{5} e^{4 - x^2} \, dx$  

usando os métodos **compostos** do **ponto médio**, **Trapézio** e **Simpson**.  
Obtenha os resultados utilizando, em cada quadratura, o número de pontos indicado.

**Respostas esperadas:**

| $n$ | Ponto médio | Trapézios | Simpson |
|:--:|:--:|:--:|:--:|
| 3 | 0.1056606 | 0.7503919 | 0.5005225 |
| 5 | 0.1726140 | 0.3964724 | 0.2784992 |
| 7 | 0.1973663 | 0.3062023 | 0.2393551 |
| 9 | 0.2084204 | 0.2721145 | 0.2306618 |

````python
import sys
import os
import numpy as np

sys.path.append(os.path.abspath(os.path.join('..')))

def f(x):
    return np.exp(4.0 - x**2)

def composto_ponto_medio(f, a, b, n_pontos):
    h = (b - a) / n_pontos
    médios = a + (np.arange(n_pontos) + 0.5) * h
    return h * np.sum(f(médios))

def composto_trapezio(f, a, b, n_pontos):
    m = n_pontos - 1
    h = (b - a) / m
    x = np.linspace(a, b, n_pontos)
    y = f(x)
    return h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])

def composto_simpson(f, a, b, n_pontos):
    m = n_pontos - 1
    h = (b - a) / m
    x = np.linspace(a, b, n_pontos)
    y = f(x)
    return (h / 3.0) * (y[0] + 4.0 * np.sum(y[1:-1:2]) + 2.0 * np.sum(y[2:-1:2]) + y[-1])

def main():
    a, b = 2.0, 5.0
    pontos_lista = [3, 5, 7, 9]

    print(f"{'n':<4} | {'Ponto médio':>14} | {'Trapézios':>14} | {'Simpson':>14}")
    print("-" * 52)
    for n in pontos_lista:
        rm = composto_ponto_medio(f, a, b, n)
        rt = composto_trapezio(f, a, b, n)
        rs = composto_simpson(f, a, b, n)
        print(f"{n:<4} | {rm:>14.7f} | {rt:>14.7f} | {rs:>14.7f}")

if __name__ == "__main__":
    main()
````
